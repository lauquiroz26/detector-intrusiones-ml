```python
import pandas as pd
import numpy as np

archivos = [
    "Monday-WorkingHours.pcap_ISCX.csv","Tuesday-WorkingHours.pcap_ISCX.csv",
    "Wednesday-workingHours.pcap_ISCX.csv","Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv",
    "Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX.csv","Friday-WorkingHours-Morning.pcap_ISCX.csv",
    "Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX.csv","Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv"]
# ------------------------------------------------
#2. LIMPIEZA
# ------------------------------------------------

dfs_limpios = {}
dataframes = []

for archivo in archivos:

    print("\n" + "=" * 70)
    print(archivo)
    print("=" * 70)

    #cargar
    df = pd.read_csv(archivo)

    #limpiar espacios de los nombres de columnas
    df.columns = df.columns.str.strip()

    filas_antes = len(df)

    #convertir infinitos a nan
    df["Flow Bytes/s"] = df["Flow Bytes/s"].replace([np.inf, -np.inf],np.nan)

    #eliminar filas con valores faltantes en Flow Bytes/s
    df = df.dropna(subset=["Flow Bytes/s"])

    filas_despues_nan = len(df)

    #eliminar duplicados exactos
    duplicados = df.duplicated().sum()
    df = df.drop_duplicates()

    filas_despues = len(df)

    # Guardar DataFrame limpio
    #dfs_limpios[archivo] = df

    print(f"Filas iniciales:        {filas_antes:,}")
    print(f"Después de NaN:         {filas_despues_nan:,}")
    print(f"Duplicados eliminados:  {duplicados:,}")
    print(f"Filas finales:          {filas_despues:,}")

    #guardar nombre del archivo
    df["Source_File"] = archivo

    #guardar en la lista
    dataframes.append(df)

#-----------------------
#2.5. UNION DE DATASETS
#-----------------------

df_total = pd.concat(dataframes, ignore_index=True)
print("\n" + "=" * 70)
print("Filas totales:", len(df_total))
print("Columnas totales:", len(df_total.columns))

#corregir simbolos
df_total["Label"] = df_total["Label"].replace({
    "Web Attack � Brute Force": "Web Attack - Brute Force",
    "Web Attack � XSS": "Web Attack - XSS",
    "Web Attack � Sql Injection": "Web Attack - Sql Injection"})

df_total["Target"] = df_total["Label"].apply(lambda x: 0 if x == "BENIGN" else 1)

```

```python

# ------------------------------------------------
#4. MACHINE LEARNING
# ------------------------------------------------
#orden de los datos
orden_dias = [
    "Monday-WorkingHours.pcap_ISCX.csv",
    "Tuesday-WorkingHours.pcap_ISCX.csv",
    "Wednesday-workingHours.pcap_ISCX.csv",
    "Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv",
    "Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX.csv",
    "Friday-WorkingHours-Morning.pcap_ISCX.csv",
    "Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX.csv",
    "Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv"]

df_total["Dia_Orden"] = df_total["Source_File"].map({
    archivo: i for i, archivo in enumerate(orden_dias)})

df_total = df_total.sort_values("Dia_Orden").reset_index(drop=True)

print(df_total["Source_File"].value_counts().sort_index())

X = df_total.drop(columns=["Target", "Label", "Source_File"])

y = df_total["Target"]

# ------------------------------------------------
# SEPARACIÓN TEMPORAL POR TRAIN, VALIDATION Y TEST
# ------------------------------------------------

train_archivos = [
    "Monday-WorkingHours.pcap_ISCX.csv",
    "Tuesday-WorkingHours.pcap_ISCX.csv",
    "Wednesday-workingHours.pcap_ISCX.csv",
    "Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv",
    "Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX.csv"]

val_archivos = [
    "Friday-WorkingHours-Morning.pcap_ISCX.csv"]

test_archivos = [
    "Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX.csv",
    "Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv"]

train = df_total[df_total["Source_File"].isin(train_archivos)]
val = df_total[df_total["Source_File"].isin(val_archivos)]
test = df_total[df_total["Source_File"].isin(test_archivos)]

print("\nTamaño de los conjuntos:")
print("Train:", train.shape)
print("Validation:", val.shape)
print("Test:", test.shape)

# ----------------
# VARIABLES X E Y
# ----------------

columnas_no_predictoras = [
    "Target",
    "Label",
    "Source_File",
    "Dia_Orden"]

X_train = train.drop(columns=columnas_no_predictoras)
y_train = train["Target"]

X_val = val.drop(columns=columnas_no_predictoras)
y_val = val["Target"]

X_test = test.drop(columns=columnas_no_predictoras)
y_test = test["Target"]
```

```python
# ------------------------------------------------
# 5. BASELINE
# ------------------------------------------------

from sklearn.dummy import DummyClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix


#
X_dummy_train = np.zeros((len(y_train), 1))
X_dummy_val = np.zeros((len(y_val), 1))

# entrenar modelo
modelo_baseline = DummyClassifier(
    strategy="most_frequent"
)

modelo_baseline.fit(X_dummy_train, y_train)

# predecir sobre la validación temporal
y_pred_baseline = modelo_baseline.predict(X_dummy_val)

print("MATRIZ DE CONFUSIÓN - BASELINE")
print(confusion_matrix(y_val, y_pred_baseline))

print("\nREPORTE DE CLASIFICACIÓN - BASELINE")
print(classification_report(
    y_val,
    y_pred_baseline,
    zero_division=0
))

#liberar matrices
del X_dummy_train, X_dummy_val
```

```python
# ------------------------------------------------
# 6. REGRESIÓN LOGÍSTICA
# ------------------------------------------------

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.model_selection import train_test_split

# 6.1 separar entrenamiento y validacion

X_train_modelo, X_val_modelo, y_train_modelo, y_val_modelo = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    stratify=y_train,
    random_state=42)

# 6. tomar muestra de 200.000 solo en train
X_train_muestra, _, y_train_muestra, _ = train_test_split(
    X_train_modelo,
    y_train_modelo,
    train_size=200000,
    stratify=y_train_modelo,
    random_state=42)

print("Entrenamiento para modelos:", X_train_muestra.shape)
print("Validación:", X_val_modelo.shape)

print("\nClases de entrenamiento:")
print(y_train_muestra.value_counts())

print("\nClases de validación:")
print(y_val_modelo.value_counts())


# 6.3 escalar las variables

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_muestra_scaled = scaler.fit_transform(
    X_train_muestra
)

X_val_modelo_scaled = scaler.transform(
    X_val_modelo
)

print("Datos escalados correctamente.")
print("Train:", X_train_muestra_scaled.shape)
print("Validation:", X_val_modelo_scaled.shape)

# 6.4 Entrenar el modelo
modelo_logistico = LogisticRegression(
    max_iter=1000,
    class_weight="balanced",
    n_jobs=-1
)

modelo_logistico.fit(
    X_train_muestra_scaled,
    y_train_muestra
)

print("\nModelo entrenado correctamente.")

# 6.5 predicciones
y_pred_logistico = modelo_logistico.predict(
    X_val_modelo_scaled
)

# 6.6 evaluar resultados
print("\nMATRIZ DE CONFUSIÓN")
print(
    confusion_matrix(
        y_val_modelo,
        y_pred_logistico
    )
)

print("\nREPORTE DE CLASIFICACIÓN")
print(
    classification_report(
        y_val_modelo,
        y_pred_logistico,
        zero_division=0
    )
)
```

```python
# ------------------------------------------------
#7. RANDOM FOREST
# ------------------------------------------------

#7.1 entrenar modelo
from sklearn.ensemble import RandomForestClassifier

# 7.1 Entrenar el modelo
modelo_rf = RandomForestClassifier(
    n_estimators=100,
    max_depth=15,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

modelo_rf.fit(
    X_train_muestra,
    y_train_muestra
)

print("Random Forest entrenado correctamente.")

# 7.2 predicciones
y_pred_rf = modelo_rf.predict(X_val_modelo)

# 7.3 evaluar resultados
print("\nMATRIZ DE CONFUSIÓN")
print(
    confusion_matrix(
        y_val_modelo,
        y_pred_rf
    )
)

print("\nREPORTE DE CLASIFICACIÓN")
print(
    classification_report(
        y_val_modelo,
        y_pred_rf,
        zero_division=0
    )
)
```

```python
# ------------------------------------------------
# 8. COMPARACION DE MODELOS Y PR-AUC (average decision)
# ------------------------------------------------

from sklearn.metrics import (precision_score, recall_score, average_precision_score)


# probabilidad de ataque
prob_logistico = modelo_logistico.predict_proba(X_val_modelo_scaled)[:, 1]

prob_rf = modelo_rf.predict_proba(X_val_modelo)[:, 1]

#regresion logistica metrics
precision_log = precision_score(
    y_val_modelo, y_pred_logistico, zero_division=0)
recall_log = recall_score(
    y_val_modelo, y_pred_logistico, zero_division=0)
pr_auc_log = average_precision_score(
    y_val_modelo, prob_logistico)

#random forest metrics
precision_rf = precision_score(
    y_val_modelo, y_pred_rf, zero_division=0)
recall_rf = recall_score(
    y_val_modelo, y_pred_rf, zero_division=0)
pr_auc_rf = average_precision_score(
    y_val_modelo, prob_rf)

# tabla comparativa
comparacion = pd.DataFrame({
    "Modelo": ["Regresion Logistica", "Random Forest"],
    "Precision": [precision_log, precision_rf],
    "Recall": [recall_log, recall_rf],
    "PR-AUC": [pr_auc_log, pr_auc_rf]
})
print(comparacion.round(4).to_string(index=False))


```

```python

# -----------------------------------------
# 9. EVALUACIÓN TEMPORAL DE RANDOM FOREST
# -----------------------------------------

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    average_precision_score
)

# 9.1 Predecir las clases del conjunto de prueba
y_pred_test_rf = modelo_rf.predict(X_test)

# 9.2 Obtener las probabilidades de ataque
prob_test_rf = modelo_rf.predict_proba(X_test)[:, 1]

# 9.3 Mostrar la matriz de confusión
print("MATRIZ DE CONFUSIÓN - PRUEBA TEMPORAL")
print(confusion_matrix(y_test, y_pred_test_rf))

# 9.4 Mostrar las métricas por clase
print("\nREPORTE DE CLASIFICACIÓN")
print(classification_report(
    y_test,
    y_pred_test_rf,
    zero_division=0
))

# 9.5 Calcular métricas principales
precision_test_rf = precision_score(
    y_test, y_pred_test_rf, zero_division=0
)

recall_test_rf = recall_score(
    y_test, y_pred_test_rf, zero_division=0
)

pr_auc_test_rf = average_precision_score(
    y_test, prob_test_rf
)

print("\nMÉTRICAS DE PRUEBA TEMPORAL")
print(f"Precision: {precision_test_rf:.4f}")
print(f"Recall:    {recall_test_rf:.4f}")
print(f"PR-AUC:    {pr_auc_test_rf:.4f}")
```

```python

# -----------------------------------------
# 10. RECALL POR TIPO DE ATAQUE
# -----------------------------------------

#10.1 Conservar las etiquetas reales del conjunto de prueba
resultado_test = test[["Label", "Target"]].copy()

#10.2 Añadir las predicciones de Random Forest
resultado_test["Prediccion"] = y_pred_test_rf

#10.3 Calcular el recall de cada tipo de ataque
resultados_por_tipo = []

for etiqueta, grupo in resultado_test.groupby("Label"):

    if grupo["Target"].iloc[0] == 1:

        recall_tipo = recall_score(
            grupo["Target"],
            grupo["Prediccion"],
            zero_division=0
        )

        resultados_por_tipo.append({
            "Tipo de ataque": etiqueta,
            "Cantidad": len(grupo),
            "Recall": recall_tipo
        })

#10.4 Crear y mostrar la tabla
tabla_recall = pd.DataFrame(resultados_por_tipo)

print("RECALL POR TIPO DE ATAQUE")
print(
    tabla_recall
    .sort_values("Recall")
    .round(4)
    .to_string(index=False))
```

```python

# -----------------------------------------
#11. RANKING DE LAS 40 ALERTAS PRIORITARIAS
# -----------------------------------------

# copiar los datos
ranking_alertas = test.copy()

#añadir la prob
ranking_alertas["Riesgo_Ataque"] = prob_test_rf

#ordenar de mayor a menor riesgo
ranking_alertas = ranking_alertas.sort_values(
    by="Riesgo_Ataque",
    ascending=False
).reset_index(drop=True)

#seleccionar las 40 alertas prioritarias
top_40 = ranking_alertas.head(40).copy()

#mostrar las columnas
columnas_mostrar = [
    "Label",
    "Riesgo_Ataque"
]

print("TOP 40 ALERTAS PRIORITARIAS")
print(
    top_40[columnas_mostrar]
    .to_string(index=False)
)
```

```python

# -----------------------------------------
# 12. VARIABLES MÁS IMPORTANTES
# -----------------------------------------

importancias = pd.DataFrame({
    "Variable": X_train_muestra.columns,
    "Importancia": modelo_rf.feature_importances_
})

importancias = importancias.sort_values(
    by="Importancia",
    ascending=False
).head(10)

print("TOP 10 VARIABLES MÁS IMPORTANTES")
print(importancias.to_string(index=False))
```

```python

# -----------------------------------------
# 13. RESUMEN FINAL DE RESULTADOS
# -----------------------------------------

resumen_final = pd.DataFrame({
    "Metrica": [
        "Precision en prueba temporal",
        "Recall en prueba temporal",
        "PR-AUC en prueba temporal",
        "Recall de PortScan",
        "Recall de DDoS"
    ],
    "Resultado": [
        precision_test_rf,
        recall_test_rf,
        pr_auc_test_rf,
        tabla_recall.loc[
            tabla_recall["Tipo de ataque"] == "PortScan",
            "Recall"
        ].iloc[0],
        tabla_recall.loc[
            tabla_recall["Tipo de ataque"] == "DDoS",
            "Recall"
        ].iloc[0]
    ]
})

print("RESUMEN FINAL DEL MODELO")
print(resumen_final.to_string(index=False, formatters={
    "Resultado": lambda x: f"{x:.4f}"
}))
```
